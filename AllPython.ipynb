# OOPs concept in python
1. Inheritance:
    Inheritance is a fundamental concept in OOPs that allows a class to inherit attributes and methods from another class.


```python
class A():
  a = " Hello, I am Apple!"

  def a(self):
    print(self.a)

class B():
  b = "Hello, I am Boll!"

  def b(self):
    print(self.b)

class C(A,B):

  c = " Hello, I am Cute!"

  def c(self):
    print(self.c)


# Instantiate objects and call methods after class definitions
objA = A()
objB = B()
objC = C()

objA.a()
objB.b()
objC.c()
```

# The __init__() Method
All classes have a built-in method called __init__(), which is always executed when the class is being initiated.

The __init__() method is used to assign values to object properties, or to perform operations that are necessary when the object is being created.

```python
class student:
  name = "RAj"
  def __init__(Self):
    print(Self.name)
obj = student()
```

# Polymorphism
Polymorphism is often used in Class methods, where we can have multiple classes with the same method name.

```python
# Example overloading code
class Ws:
  def displayinfo(self,name=''):
    print("Hello I am  " , name)
obj = Ws()
obj = Ws()
obj.displayinfo()
obj.displayinfo("Raj")
```

# Question

The super() function is used to give access to methods and properties of a parent or sibling class.

The super() function returns an object that represents the parent class.



```python
class Ws:
  def displayinfo(self,name=''):
    print("Hello my name is a " , name)
class IP(Ws):
  def displayinfo(self):
    super().displayinfo("raj")
    print("Welcome to ip")
obj= IP()
obj.displayinfo()

```

#  Getter and Setter in Python


```python
class S:
  def __init__(Self):
    Self.__name=""
  def getname(Self):
    return Self.__name
  def setname(Self, name):
    Self.__name = name

obj = S()
obj.setname( "RAJ")
name = obj.getname()
print(name)
```
