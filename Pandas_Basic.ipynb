Adding, Updating & Deleting Columns

```python
# 1. Add a new column
# Why: Existing data se ek new feature create karne ke liye
df["Passed"] =df ["Marks"] >=40

print("\nAfter adding Passed column:")
print(df)
```

# 2. Add a calculated column
# Why: Existing numerical data se new calculated feature banane ke liye

```python
df["Marks_Double"] = df["Marks"] * 2

print("After adding a Marks_Double column:")
print(df)
```

# 3. Add another calculated column
# Why: Future value calculate karke new feature banane ke liye

```python
df["Age_Next_Year"] = df["Age"] + 1

print("\n after adding a Age_Next_yera column:\n")
print(df)
```

# 4. Conditional update
### Why: Condition ke basis par specific values modify karne ke liye

```python
df.loc[df["Marks"] < 40, "Marks"] = 40

print("\nAfter updating Marks below 40:")
print(df)
```

# 5. Delete one column
### Why: Unwanted column remove karne ke liye

```python
df.drop("Marks_Double", axis=1, inplace=True)

print("\nAfter deleting Marks_Double:")
print(df)
```

# 6. Delete multiple columns
### Why: Ek saath multiple unwanted columns remove karne ke liye

```python
df.drop(["Age_Next_Year"], axis=1, inplace=True)

print("\nAfter deleting Age_Next_Year:")
print(df)
```

#Sorting Data

# 1. Sort by Marks in ascending order.

```python
df_sorted_marks = df.sort_values("Marks")

print("\nMarks - Ascending Order:")
print(df_sorted_marks)
```

# 2. Sort by Marks in descending order


```python
df_sorted_marks_desc = df.sort_values("Marks", ascending=False)

print("\nMarks - Descending Order:")
print(df_sorted_marks_desc)
```

# 3. Sort by Age


```python
df_sorted_age = df.sort_values("Age")

print("\nAge - Ascending Order:")
print(df_sorted_age)
```

# 4. Sort by Age in descending order

```python
df_sorted_age_desc = df.sort_values("Age", ascending=False)

print("\nAge - Descending Order:")
print(df_sorted_age_desc)
```

# 5. Sort by multiple columns


```python
df_sorted_multiple = df.sort_values(
    ["Age", "Marks"],
    ascending=[True, False]
)

print("\nSort by Age Ascending and Marks Descending:")
print(df_sorted_multiple)
```

# 6. Sort by Name


```python
df_sorted_name = df.sort_values("Name")

print("\nName - Alphabetical Order:")
print(df_sorted_name)
```
