```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Name: Raj Mishra")
df = sns.load_dataset("titanic")
print("Dataset loaded successfully!")
```

## Q1. View the Data (1 mark)

```python
print(df.head())
print("Number of rows and columns:", df.shape)
```

## Q2. Understand the Data (1 mark)

```python
df.info()

print("\nSummary statistics of numerical columns:")
print(df.describe())
```

## Q3. Handle Missing Values (1 mark)

```python
print("Missing values in each column:")
print(df.isnull().sum())

median_age = df["age"].median()
print("\nMedian age:", median_age)

df["age"] = df["age"].fillna(median_age)

print("\nMissing values in age after filling:")
print(df["age"].isnull().sum())
```

## Q4. Visualize the Data (1 mark)

```python
# (a) Histogram of age
sns.histplot(df["age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# (b) Count plot of survived
sns.countplot(x="survived", data=df)
plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```
